```python
import os
import soundfile as sf
from collections import defaultdict

def summarize_audio_folder(folder_path):
    patient_durations = defaultdict(list)

    for file in os.listdir(folder_path):
        if file.endswith(".wav"):
            patient_id = file.split("_segment")[0]
            file_path = os.path.join(folder_path, file)

            data, samplerate = sf.read(file_path)
            duration = len(data) / samplerate
            patient_durations[patient_id].append(duration)

    # header
    print("\nPatient-wise summary:")
    print(f"{'Patient':<15} {'Segments':<10} {'Avg (s)':<10} {'Total (s)':<10}")
    print("-" * 50)

    for patient_id, durations in sorted(patient_durations.items()):
        num_files = len(durations)
        avg_duration = sum(durations) / num_files
        total_duration = sum(durations)

        print(f"{patient_id:<15} "
              f"{num_files:<10} "
              f"{avg_duration:<10.2f} "
              f"{total_duration:<10.2f}")

    # overall summary
    patient_total_durations = [sum(d) for d in patient_durations.values()]

    if patient_total_durations:
        print("\nOverall summary:")
        print(f"Total patients: {len(patient_total_durations)}")
        print(f"Average total duration per patient: "
              f"{sum(patient_total_durations) / len(patient_total_durations):.2f} s")


summarize_audio_folder("../data/Spontanaktivität")
```
